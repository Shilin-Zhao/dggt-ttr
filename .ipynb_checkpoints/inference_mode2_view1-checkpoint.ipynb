```python
# -----------------------------------------------------------------------------
# 1. System & Path Setup
# -----------------------------------------------------------------------------
import os
if 'OMP_NUM_THREADS' in os.environ:
    del os.environ['OMP_NUM_THREADS']
import sys
import torch
import matplotlib.pyplot as plt
from tqdm.notebook import tqdm
os.environ['HF_HUB_OFFLINE'] = '1'
os.environ['HF_HOME'] = os.path.expanduser('~/.cache/huggingface')
print("Hugging Face offline mode enabled.")

# -----------------------------------------------------------------------------
# 2. Project Imports 

# -----------------------------------------------------------------------------
from core_pipeline import *
from vis_utils import *
from datasets.dataset import WaymoOpenDataset

# -----------------------------------------------------------------------------
# 3. Environment Check
# -----------------------------------------------------------------------------
%matplotlib inline

device = "cuda" if torch.cuda.is_available() else "cpu"
print(f"Environment ready. Device: {device}")
```

```python
# -----------------------------------------------------------------------------
# 实例化 Config (对应你提供的命令行参数)
# -----------------------------------------------------------------------------
scene_names = ["000"]
input_views = 1
squence_length = 20
start_idx=0
force_refresh = False
# output_path=f"outputs/scene{scene_name}_start{start_idx}_len{squence_length}_view{input_views}"

config = DGGTConfig(
    # 必需路径参数
    image_dir="data/waymo/processed/validation",
    ckpt_path="/root/pretrained/model_latest_waymo.pt", 
    output_path="outputs/dggt",
    
    # 场景设置
    scene_names=scene_names,       # --scene_name 000
    input_views=input_views,             # --input_views 1
    sequence_length=squence_length,        # --sequence_length 20
    start_idx=start_idx,               # --start_idx 0
    
    # 功能开关 (根据你的命令行: -images -metrics -diffusion)
    images=True,
    metrics=True,
    
    # Refinement 开关 (命令行未开启，设为False，后续调试可改为True)
    enable_refinement=False,   
    depth=True
)

print(f"Config loaded for scenes: {config.scene_names}")
print(f"Output path: {config.output_path}")
print(f"Refinement enabled: {config.enable_refinement}")
```

```python
# 实例化 Dataset
# 假设 datasets.dataset 已经可以 import 并且包含了你提供的逻辑
# 如果是在 autodl 本地文件，确保路径正确
dataset = WaymoOpenDataset(
    image_dir=config.image_dir,
    scene_names=scene_names,
    sequence_length=config.sequence_length,
    start_idx=config.start_idx,
    mode=2,             # pure reconstruction
    views=config.input_views
)
```

```python
# 运行可视化
visualize_samples(dataset, [2, 6, 12])
```

```python
# -----------------------------------------------------------------------------
# 4. Load Model, Run Inference & Visualize Results
# -----------------------------------------------------------------------------

# A simple logger for notebook use
class NotebookLogger:
    def info(self, msg):
        print(f"[INFO] {msg}")

# 1. Setup Environment & Model
logger = NotebookLogger()
device = "cuda" if torch.cuda.is_available() else "cpu"

print(f"Loading checkpoint from: {config.ckpt_path}")
model = load_model(config, device, logger)
```

```python
print("Extracting depth data...")
dggt_depth, dggt_sky, da3_depth, da3_sky = extract_depth_alignment_data(config, model, dataset, device)

run_depth_alignment_debug(
    sample_indices=[2, 6, 12],
    dggt_depth_seq=dggt_depth,
    dggt_sky_seq=dggt_sky,
    da3_depth_seq=da3_depth,
    da3_sky_seq=da3_sky,
    dataset=dataset,
    logger=logger,
    visualize_func=visualize_depth_alignment_6panel
)
```

```python
# 2. Run Inference
# Ensure config matches your current memory constraints (seq_len=2)
print(f"Running inference for scene: {config.scene_names} (Seq Len: {config.sequence_length})...")

# results is a list of dicts (one per scene)
config.diffusion =  False
config.penalize_alphas = False
config.use_nearest = True
config.metrics = True
config.enable_refinement = False
config.refine_pose = True
config.refine_gs = True
config.enable_da3 = True
results = run_inference_and_render(config, model, dataset, logger, device=device, return_ply=True)
scene_data = results[0] # Get first scene
```

```python
# 2.1 Export 4D Gaussians to PLY (static / dynamic split)
# logical frame index: e.g. frame 0 and 5
ply_payload = scene_data.get("ply_payload", None)
if ply_payload is None:
    raise RuntimeError("No ply_payload found. Please run inference with return_ply=True.")

ply_output_dir = os.path.join(config.output_path, "da3_dggt_intersect", "ply_exports")
exported_info = save_gaussians_payload_to_ply(
    ply_payload=ply_payload,
    frame_indices=[0, 5],
    output_dir=ply_output_dir,
    scene_name=scene_data["scene_name"],
    logger=logger,
    min_opacity=.0
)

print("PLY export done:")
for item in exported_info:
    print(item)
```

```python
# -----------------------------------------------------------------------------
# 9. Full Sequence Inference with Overlap Strategy
# -----------------------------------------------------------------------------
# 核心函数: process_single_scene (core_pipeline.py)
# - 处理单场景的完整序列推理
# - 自动计算 metrics
# - 自动保存视频和帧图片



# 配置参数
config.output_path = "outputs/dggt_ttr"
config.diffusion = True
config.penalize_alphas = False
config.use_nearest = True
config.enable_refinement = True
config.metrics = True  # 启用 metrics 计算
config.refine_pose = True
config.refine_gs = True
config.refine_xyz = True

# 输出目录
output_base_dir = config.output_path
log_dir = os.path.join(output_base_dir, "logs")
os.makedirs(log_dir, exist_ok=True)

# 结果存储
results = {}
all_metrics = []
scene_names=["007","005","004","003","001","000"][::-1]
# scene_names=["000"]
# 循环处理每个场景
for scene_name in tqdm(scene_names, desc="🎬 Scenes"):
    print(f"\n{'='*60}")
    print(f"📍 Scene: {scene_name}")
    print(f"{'='*60}")
    
    # 创建场景专属的 logger（只写文件，不打印）
    scene_logger = FileLogger(log_dir, scene_name)
    scene_logger.info(f"Starting inference for scene {scene_name}")
    
    # 处理单个场景
    result = process_single_scene(
        config=config,
        model=model,
        dataset_class=WaymoOpenDataset,
        scene_name=scene_name,
        device=device,
        sequence_length=20,
        overlap=3,
        output_dir=output_base_dir,
        save_video_func=save_full_sequence_video,
        fps=8,
        logger=scene_logger
    )
    
    results[scene_name] = result
    scene_logger.close()
    
    # 打印场景结果
    if result['video_path'] is not None:
        print(f"✅ Video: {result['video_path']}")
        print(f"   Frames: {result['num_frames']}")
        if result['metrics'] is not None:
            m = result['metrics']
            print(f"   PSNR:  {m['psnr']:.4f}")
            print(f"   SSIM:  {m['ssim']:.4f}")
            print(f"   LPIPS: {m['lpips']:.4f}")
            all_metrics.append(m)
    else:
        print(f"❌ FAILED")

# 汇总统计
print("\n" + "="*60)
print("📊 Summary")
print("="*60)

if all_metrics:
    import numpy as np
    avg_psnr = np.mean([m['psnr'] for m in all_metrics])
    avg_ssim = np.mean([m['ssim'] for m in all_metrics])
    avg_lpips = np.mean([m['lpips'] for m in all_metrics])
    
    print(f"Average Metrics ({len(all_metrics)} scenes):")
    print(f"  PSNR:  {avg_psnr:.4f}")
    print(f"  SSIM:  {avg_ssim:.4f}")
    print(f"  LPIPS: {avg_lpips:.4f}")
else:
    print("No metrics computed.")

print(f"\n📁 Logs:   {log_dir}")
print(f"📁 Videos: {output_base_dir}")
```
